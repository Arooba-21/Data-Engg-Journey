```python
# Sample employee data (jaisa tumne pehle banaya tha)
data = [
    (1, "Ali", "IT", 90000),
    (2, "Sara", "HR", 70004),
    (3, "Adil", "AI", 90000),
    (4, "amna", "Finance", 90400),
    (5, "Hiba", "HR", 94000),
    (6, "Hina", "IT", 90004)
]

columns = ["id", "name", "Dept", "salary"]

df = spark.createDataFrame(data, columns)
df.show()
```

```python
df.write.format("delta").mode("overwrite").saveAsTable("employees_delta")
```

```python
new_data = [
    (2, "Sara", "HR", 80000),      # existing employee, salary update
    (7, "Bilal", "IT", 95000)       # naya employee
]

new_columns = ["id", "name", "Dept", "salary"]
new_df = spark.createDataFrame(new_data, new_columns)
new_df.createOrReplaceTempView("new_employee_data")
```

```python
%sql
MERGE INTO employees_delta AS target
USING new_employee_data AS source
ON target.id = source.id
WHEN MATCHED THEN
    UPDATE SET target.salary = source.salary
WHEN NOT MATCHED THEN
    INSERT (id, name, Dept, salary) VALUES (source.id, source.name, source.Dept, source.salary)
```
